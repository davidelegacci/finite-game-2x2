# AdaExtraFTRL — Regret robustness against a deviating player

This notebook extends `2026-02-13_AdaExtraFTRL_222_2222.ipynb` with a
**regret computation** and a **single, comparison plot** that quantifies the
difference between

* `constFTRLplus` — Legacci et al. et al. (NeurIPS '24): **brittle**. A single
  player deviating from the prescribed parameters can drive the regret of
  the conforming players to grow **super-linearly**.
* `adaFTRLplus` — the parameter-agnostic adaptive scheme (Eq. (12) in the
  paper): **robust**. Even with one deviating player, conforming players
  retain $\mathcal{O}(\sqrt{T})$ regret (Theorem 4(a)).

**Setup.** A 3-player 2-action ($K-N=3$) harmonic game from the parent
notebook. We run four scenarios with identical initial conditions:

| | Player 0 | Player 1 | Player 2 (deviator) |
|---|---|---|---|
| **A. const, self-play** | const, $\eta=0.05$ | const, $\eta=0.05$ | const, $\eta=0.05$ |
| **B. ada, self-play**   | ada                  | ada                  | ada                                 |
| **C. const + deviator** | const, $\eta=0.05$ | const, $\eta=0.05$ | const, $\eta=5$ (mistuned) |
| **D. ada + deviator**   | ada                  | ada                  | const, $\eta=5$ (mistuned) |

In C and D, **player 2 is identical** (same mistuned scheme). The contrast
between C and D therefore isolates the effect of the conforming players'
choice of algorithm. The deviator is "honest but ignorant" — the threshold
$\gamma_i \le \kappa_i \cdot (\min_j m_j \min_j \kappa_j) /
(16 m_i N L_i^2)$ from Eq. (11) **actually need threshold from NeurIPS 2024 paper**depends on harmonic masses, Lipschitz
moduli, and the player count, none of which an individual player can know,
so picking $\eta$ from a "plausible" range can easily fall outside the
safe regime.

```python
u_tensor = nfg.Utils.u_flat_to_tensor(u_flat, skeleton)

pure_labels = [['B', 'F'], ['L', 'R'], ['D', 'U']]

game = nfg.NFG(skeleton, 'literal', 'x', 'dual', '222-harmonic',
               u_tensor, pure_labels=pure_labels)
```

```python
vas = game.letters[:game.num_players]
print('symbolic variables:', vas)
```

## Effective payfield

```python
payfield = sp.lambdify(vas, game.clean_payfield)

# Sanity check: evaluate at barycenter
BARYCENTER = [0.5 for _ in range(EFFECTIVE_DIMENSION)]
print('payfield at barycenter:', payfield(*BARYCENTER))
```

## Mirror map and signal blend

```python
# Logistic mirror -- the [0,1] specialization of the entropic mirror map.
# Numerically stable form (avoids overflow of exp for large |y|).

def mirrori(y):
    y = np.asarray(y, dtype=float)
    out = np.empty_like(y)
    pos = y >= 0
    out[pos]  = 1.0 / (1.0 + np.exp(-y[pos]))
    out[~pos] = np.exp(y[~pos]) / (1.0 + np.exp(y[~pos]))
    return out


def make_signal(coef, v_curr, v_beforelead):
    """Blend between extra-gradient (coef=1) and optimistic (coef=0)."""
    return coef * v_curr + (1 - coef) * v_beforelead
```

## Unified per-player FTRL+

A single algorithm in which each player picks **independently** between
`'const'` (constant $\eta_i$, Legacci et al.'s scheme) and `'ada'`
(parameter-agnostic adaptive policy of Eq. (12)). This subsumes both
`adaFTRLplus` and `constFTRLplus` from the parent notebook and lets us
mix them per-player in the same run.


```python
def perPlayerFTRLplus(y_1, nRuns, schemes, etas, coefs, v_half=None):
    """
    Run FTRL+ where each player has its own learning-rate scheme.

    Parameters
    ----------
    y_1     : array (EFFECTIVE_DIMENSION,)  initial dual point
    nRuns   : int                            number of iterations
    schemes : sequence of {'const', 'ada'}   per-player scheme
    etas    : array (EFFECTIVE_DIMENSION,)   constant eta ('const')
                                             or initial eta ('ada')
    coefs   : array (EFFECTIVE_DIMENSION,)   extrapolation coefficients lambda_i
    v_half  : array, optional                initial pseudo-payoff signal v_{1/2}

    Returns
    -------
    lead_sequence : array (nRuns, EFFECTIVE_DIMENSION) of x_{t+1/2}
    """
    y_1   = np.asarray(y_1,   dtype=float)
    etas  = np.asarray(etas,  dtype=float).copy()
    coefs = np.asarray(coefs, dtype=float)
    schemes = np.asarray(schemes)
    is_ada = (schemes == 'ada')

    if v_half is None:
        v_half = np.zeros(EFFECTIVE_DIMENSION)

    # Initialization
    x_curr      = mirrori(etas * y_1)
    v_curr      = np.asarray(payfield(*x_curr), dtype=float)
    signal_curr = make_signal(coefs, v_curr, v_half)
    y_curr      = y_1.copy()
    eta_curr    = etas.copy()
    g_sum_curr  = np.zeros(EFFECTIVE_DIMENSION)

    lead_sequence = np.empty((nRuns, EFFECTIVE_DIMENSION))

    for t in range(nRuns):
        # Extrapolation
        y_lead = y_curr + signal_curr
        x_lead = mirrori(eta_curr * y_lead)
        lead_sequence[t] = x_lead
        v_lead = np.asarray(payfield(*x_lead), dtype=float)

        # Difference of gradients
        g_curr      = v_lead - signal_curr
        g_sum_curr += g_curr * g_curr

        # Per-player learning-rate update
        eta_next_ada = (1.0 + g_sum_curr) ** (-0.5)
        eta_next     = np.where(is_ada, eta_next_ada, etas)

        # Update step
        y_next      = y_curr + v_lead
        x_next      = mirrori(eta_next * y_next)
        v_next      = np.asarray(payfield(*x_next), dtype=float)
        signal_next = make_signal(coefs, v_next, v_lead)

        # Roll
        y_curr      = y_next
        eta_curr    = eta_next
        signal_curr = signal_next

    return lead_sequence
```

### Sanity check

For purposes of equivalence, when all players use the same scheme and
parameters, `perPlayerFTRLplus` reproduces the original `adaFTRLplus`
and `constFTRLplus` from the parent notebook.


```python
# Original methods, copy-pasted for the sanity check.
# (Use mirrori as defined above, which is the stable variant.)

def adaFTRLplus(y_1, nRuns,
                eta_1=np.ones(EFFECTIVE_DIMENSION),
                v_half=np.zeros(EFFECTIVE_DIMENSION),
                coef=0.5 * np.ones(EFFECTIVE_DIMENSION)):
    x_1 = mirrori(eta_1 * y_1)
    v_1 = payfield(*x_1)
    signal_1 = make_signal(coef, v_1, v_half)
    t = 1
    y_curr = y_1
    eta_curr = eta_1
    signal_curr = signal_1
    g_sum_curr = np.zeros(EFFECTIVE_DIMENSION)
    lead_sequence = []
    while t <= nRuns:
        y_lead = y_curr + signal_curr
        x_lead = mirrori(eta_curr * y_lead)
        lead_sequence.append(x_lead)
        v_lead = payfield(*x_lead)
        g_curr = v_lead - signal_curr
        g_sum_curr += g_curr * g_curr
        eta_next = (1 + g_sum_curr) ** (-0.5)
        y_next = y_curr + v_lead
        x_next = mirrori(eta_next * y_next)
        v_next = payfield(*x_next)
        signal_next = make_signal(coef, v_next, v_lead)
        t += 1
        y_curr = y_next
        eta_curr = eta_next
        signal_curr = signal_next
    return np.array(lead_sequence)


def constFTRLplus(y_1, nRuns,
                  eta_1=0.1 * np.ones(EFFECTIVE_DIMENSION),
                  v_half=np.zeros(EFFECTIVE_DIMENSION),
                  coef=0.5 * np.ones(EFFECTIVE_DIMENSION)):
    x_1 = mirrori(eta_1 * y_1)
    v_1 = payfield(*x_1)
    signal_1 = make_signal(coef, v_1, v_half)
    t = 1
    y_curr = y_1
    eta_curr = eta_1
    signal_curr = signal_1
    g_sum_curr = np.zeros(EFFECTIVE_DIMENSION)
    lead_sequence = []
    while t <= nRuns:
        y_lead = y_curr + signal_curr
        x_lead = mirrori(eta_curr * y_lead)
        lead_sequence.append(x_lead)
        v_lead = payfield(*x_lead)
        g_curr = v_lead - signal_curr
        g_sum_curr += g_curr * g_curr
        eta_next = eta_curr
        y_next = y_curr + v_lead
        x_next = mirrori(eta_next * y_next)
        v_next = payfield(*x_next)
        signal_next = make_signal(coef, v_next, v_lead)
        t += 1
        y_curr = y_next
        eta_curr = eta_next
        signal_curr = signal_next
    return np.array(lead_sequence)


# Sanity check
y0 = np.array([1.0, -2.0, 0.5])
nR = 200
coefs = 0.5 * np.ones(EFFECTIVE_DIMENSION)

s_ada_ref  = adaFTRLplus(y0, nR, eta_1=np.ones(EFFECTIVE_DIMENSION),
                         v_half=np.zeros(EFFECTIVE_DIMENSION), coef=coefs)
s_ada_new  = perPlayerFTRLplus(y0, nR,
                               schemes=['ada'] * EFFECTIVE_DIMENSION,
                               etas=np.ones(EFFECTIVE_DIMENSION),
                               coefs=coefs)
print('ada match:',  np.allclose(s_ada_ref, s_ada_new))

s_const_ref = constFTRLplus(y0, nR, eta_1=0.1 * np.ones(EFFECTIVE_DIMENSION),
                            v_half=np.zeros(EFFECTIVE_DIMENSION), coef=coefs)
s_const_new = perPlayerFTRLplus(y0, nR,
                                schemes=['const'] * EFFECTIVE_DIMENSION,
                                etas=0.1 * np.ones(EFFECTIVE_DIMENSION),
                                coefs=coefs)
print('const match:', np.allclose(s_const_ref, s_const_new))
```

## Cumulative regret

For player $i \in \mathcal{N}$ in a 2-action effective representation
$x_i \in [0,1]$, the **effective payfield**
$v_i(x) = u_i(\mathrm{action\,1}; x_{-i}) - u_i(\mathrm{action\,0}; x_{-i})$
encodes everything: the mixed payoff is
$u_i(x) = u_i(0; x_{-i}) + x_i\,v_i(x)$, and a unilateral deviation to
$p_i \in [0,1]$ changes the payoff by $(p_i - x_{i,t})\,v_i(x_t)$.
Cumulative regret against the best fixed action is therefore

$$
\mathrm{Reg}_i(T)
\;=\;
\max_{p_i \in [0,1]} \sum_{t=1}^{T} (p_i - x_{i,t})\,v_i(x_t)
\;=\;
\bigl[\textstyle\sum_t v_i(x_t)\bigr]_{+}
\;-\;
\sum_t x_{i,t}\,v_i(x_t),
$$

where $[\,\cdot\,]_+$ is the positive part (the maximum is attained at
a vertex $p_i \in \{0, 1\}$).


```python
def cumulative_regret(lead_sequence):
    """Cumulative regret per player along the lead sequence.

    Parameters
    ----------
    lead_sequence : array (T, EFFECTIVE_DIMENSION)

    Returns
    -------
    regret : array (T, EFFECTIVE_DIMENSION)
        regret[t-1, i] = Reg_i over rounds 1, ..., t.
    """
    T = len(lead_sequence)
    cum_v  = np.zeros(EFFECTIVE_DIMENSION)
    cum_xv = np.zeros(EFFECTIVE_DIMENSION)
    regret = np.empty((T, EFFECTIVE_DIMENSION))
    for t, x_t in enumerate(lead_sequence):
        v_t      = np.asarray(payfield(*x_t), dtype=float)
        cum_v   += v_t
        cum_xv  += np.asarray(x_t, dtype=float) * v_t
        regret[t] = np.maximum(cum_v, 0.0) - cum_xv
    return regret
```

## Run the four scenarios

Same initial dual point $y_1$ and same number of rounds for all four runs;
the only differences are the per-player schemes and step-sizes.

We use one **deviator** (player index 2 by default) running `constFTRLplus`
with a deliberately oversized $\eta = 5$. The conformers use a
conservative $\eta = 0.05$ for the `const` scheme — well inside the safe
regime when everyone plays along.


```python
# Reproducible setup
#np.random.seed(7)

nRuns = 8000
y_1   = np.array([1.0, -2.0, 0.5])
coefs = 0.5 * np.ones(EFFECTIVE_DIMENSION)

ETA_TUNED    = 0.05  # well-tuned constant eta for Legacci et al.'s scheme (conservative)
ETA_MISTUNED = 5.0   # deviator's mistuned eta -- 100x the conformers' value
DEV_ID       = 2     # deviator is player 2 (others: 0, 1)

# A. All const, well-tuned (Legacci et al. self-play)
seq_A = perPlayerFTRLplus(
    y_1, nRuns,
    schemes=['const', 'const', 'const'],
    etas=[ETA_TUNED, ETA_TUNED, ETA_TUNED],
    coefs=coefs,
)

# B. All ada (adaptive self-play)
seq_B = perPlayerFTRLplus(
    y_1, nRuns,
    schemes=['ada', 'ada', 'ada'],
    etas=[1.0, 1.0, 1.0],   # initial eta for ada
    coefs=coefs,
)

# C. const conformers + mistuned const deviator (player DEV_ID)
schemes_C = ['const', 'const', 'const']
etas_C    = [ETA_TUNED, ETA_TUNED, ETA_TUNED]
etas_C[DEV_ID] = ETA_MISTUNED
seq_C = perPlayerFTRLplus(y_1, nRuns, schemes_C, etas_C, coefs)

# D. ada conformers + mistuned const deviator (same deviator as in C)
schemes_D = ['ada', 'ada', 'ada']
schemes_D[DEV_ID] = 'const'
etas_D = [1.0, 1.0, 1.0]
etas_D[DEV_ID] = ETA_MISTUNED
seq_D = perPlayerFTRLplus(y_1, nRuns, schemes_D, etas_D, coefs)

# Regret
reg_A = cumulative_regret(seq_A)
reg_B = cumulative_regret(seq_B)
reg_C = cumulative_regret(seq_C)
reg_D = cumulative_regret(seq_D)

CONFORMERS = [i for i in range(EFFECTIVE_DIMENSION) if i != DEV_ID]
print('Conforming players:', CONFORMERS)
print('Deviator:           ', DEV_ID)

# Worst-case (max over conformers) regret -- the relevant quantity
maxreg_A = reg_A.max(axis=1)                       # all 3 players conform
maxreg_B = reg_B.max(axis=1)                       # all 3 players conform
maxreg_C = reg_C[:, CONFORMERS].max(axis=1)        # max over conformers
maxreg_D = reg_D[:, CONFORMERS].max(axis=1)        # max over conformers


# Empirical growth-rate exponent: fit log(Reg) = c + alpha * log(T) on the
# second half of the trajectory. alpha ~ 0  =>  bounded;  alpha ~ 0.5 => sqrt(T);
# alpha ~ 1 => linear; alpha > 1 => super-linear.
def fit_growth(maxreg, start=0.5):
    T = len(maxreg)
    half = int(start * T)
    T_axis = np.arange(1, T+1)
    y = np.maximum(maxreg[half:], 1e-3)
    if np.all(y < 1e-2):
        return float('nan')
    return np.polyfit(np.log(T_axis[half:]), np.log(y), 1)[0]


print()
print('Final cumulative regret (max over conforming players):')
print(f'  A. const, self-play         : {maxreg_A[-1]:>10.2f}    growth alpha = {fit_growth(maxreg_A):>+5.2f}')
print(f'  B. ada,   self-play         : {maxreg_B[-1]:>10.2f}    growth alpha = {fit_growth(maxreg_B):>+5.2f}')
print(f'  C. const, with deviator     : {maxreg_C[-1]:>10.2f}    growth alpha = {fit_growth(maxreg_C):>+5.2f}   <-- super-linear, bound broken')
print(f'  D. ada,   with deviator     : {maxreg_D[-1]:>10.2f}    growth alpha = {fit_growth(maxreg_D):>+5.2f}   <-- robust')
```

## Regret plot

Log-log axes. The growth-rate references make the asymptotic story
explicit:

- $\mathcal{O}(1)$ — flat curve (bounded regret).
- $\mathcal{O}(\sqrt T)$ — slope $1/2$ on log-log.
- $\mathcal{O}(T)$ — slope $1$ on log-log.

**What to read off the plot.**

- A and B (self-play) hug the $\mathcal{O}(1)$ reference — both Legacci et al.
  and the adaptive scheme are bounded when everyone plays along.
- **C explodes**: with a single mistuned player, the conforming
  Legacci et al.-tuned players accumulate regret with empirical slope $\approx 1$
  (in fact often slightly above 1 in this run). The $\mathcal{O}(1)$
  guarantee is gone.
- **D stays sublinear**: the same deviator, but now the conforming
  players run the adaptive scheme. Regret stays bounded — comfortably
  inside the $\mathcal{O}(\sqrt T)$ guarantee of Theorem 4(a).


```python
SAVE = False

T_axis = np.arange(1, nRuns + 1)

# Reference curves
sqrtT      = np.sqrt(T_axis)
ref_O1     = np.full_like(T_axis, 1.0, dtype=float)        # O(1)
ref_sqrtT  = sqrtT                                          # O(sqrt T)
ref_T      = T_axis.astype(float)# / 50.0                    # O(T) (rescaled)

# Slopes for the legend
aA = fit_growth(maxreg_A); aB = fit_growth(maxreg_B)
aC = fit_growth(maxreg_C); aD = fit_growth(maxreg_D)

fig, ax = plt.subplots(figsize=(9, 6))

# Self-play (no deviator) -- dashed
ax.loglog(T_axis, np.maximum(maxreg_A, 1e-3),
          linestyle='--', color='tab:orange', linewidth=1.6,
          label=f'A. const (Legacci et al.), self-play  (slope={aA:+.2f})')
ax.loglog(T_axis, np.maximum(maxreg_B, 1e-3),
          linestyle='--', color='tab:blue',   linewidth=1.6,
          label=f'B. ada (adaptive), self-play  (slope={aB:+.2f})')

# With deviator -- solid (the contrast we care about)
ax.loglog(T_axis, np.maximum(maxreg_C, 1e-3),
          linestyle='-',  color='tab:orange', linewidth=2.6,
          label=f'C. const (Legacci et al.) + 1 deviator  (slope={aC:+.2f})')
ax.loglog(T_axis, np.maximum(maxreg_D, 1e-3),
          linestyle='-',  color='tab:blue',   linewidth=2.6,
          label=f'D. ada (adaptive) + 1 deviator  (slope={aD:+.2f})')

# References
ax.loglog(T_axis, ref_O1,     ':', color='gray',  linewidth=1.0,
          label=r'$\mathcal{O}(1)$')
ax.loglog(T_axis, ref_sqrtT,  ':', color='black', linewidth=1.0,
          label=r'$\mathcal{O}(\sqrt{T})$')
ax.loglog(T_axis, ref_T,      ':', color='red',   linewidth=1.0,
          label=r'$\mathcal{O}(T)$ (rescaled)')

ax.set_xlabel(r'horizon $T$')
ax.set_ylabel(r'cumulative regret  $\max_{i \in \mathrm{conformers}} \mathrm{Reg}_i(T)$')
ax.set_title('Robustness of FTRL+ regret to a single deviating player\n'
             '(222 harmonic game; deviator: player {} with mistuned $\\eta=$ {})'
             .format(DEV_ID, ETA_MISTUNED))
ax.grid(True, which='both', linestyle=':', alpha=0.5)
ax.legend(loc='lower right', fontsize=9)
#plt.tight_layout()
#lt.show()

if SAVE:
    out_dir = './Results/regret'
    aspera.utils.make_folder(out_dir)
    fig.savefig(f'{out_dir}/regret_robustness.pdf', bbox_inches='tight')
    print('saved to', out_dir)
```

## Per-player regret (diagnostic)

Same scenarios, but now we plot the regret of **each conforming player
separately**, so the reader can see that the failure in scenario C is not
driven by just one of the two conforming players — both blow up.


```python
fig, axs = plt.subplots(1, 2, figsize=(13, 5), sharey=True)

# Left: scenario C (Legacci et al. + deviator) -- bad
for k, p in enumerate(CONFORMERS):
    color = ['tab:orange', 'tab:red'][k]
    axs[0].loglog(T_axis, np.maximum(reg_C[:, p], 1e-3),
                  color=color, linewidth=1.8,
                  label=f'player {p} (const, conformer)')
axs[0].loglog(T_axis, ref_sqrtT, ':', color='black', label='sqrt(T) reference')
axs[0].loglog(T_axis, ref_T,     ':', color='red',   label='T reference (rescaled)')
axs[0].set_title('C. Legacci et al. tuned scheme + 1 deviator')
axs[0].set_xlabel('horizon T')
axs[0].set_ylabel('cumulative regret of player i')
axs[0].grid(True, which='both', linestyle=':', alpha=0.5)
axs[0].legend(loc='lower right', fontsize=9)

# Right: scenario D (adaptive + deviator) -- robust
for k, p in enumerate(CONFORMERS):
    color = ['tab:blue', 'tab:cyan'][k]
    axs[1].loglog(T_axis, np.maximum(reg_D[:, p], 1e-3),
                  color=color, linewidth=1.8,
                  label=f'player {p} (ada, conformer)')
axs[1].loglog(T_axis, ref_sqrtT, ':', color='black', label='sqrt(T) reference')
axs[1].loglog(T_axis, ref_T,     ':', color='red',   label='T reference (rescaled)')
axs[1].set_title('D. Adaptive scheme + 1 deviator')
axs[1].set_xlabel('horizon T')
axs[1].grid(True, which='both', linestyle=':', alpha=0.5)
axs[1].legend(loc='lower right', fontsize=9)

# subplots_adjust instead of tight_layout: tight_layout triggers a known
# matplotlib + ipympl ('%matplotlib widget') bug that invokes the mathtext
# parser on text that has already been rendered, raising a ParseException.
fig.subplots_adjust(left=0.07, right=0.97, bottom=0.12, top=0.91, wspace=0.05)
plt.show()
```

## Trajectory diagnostic

To check that the deviator is not behaving pathologically (e.g.\ stuck on
the boundary), here is each player's strategy trajectory in scenario C
and D.


```python
fig, axs = plt.subplots(1, 2, figsize=(13, 4), sharey=True)

for i in range(EFFECTIVE_DIMENSION):
    label = f'player {i}' + (' (deviator)' if i == DEV_ID else '')
    axs[0].plot(seq_C[:, i], linewidth=0.8, label=label)
    axs[1].plot(seq_D[:, i], linewidth=0.8, label=label)

axs[0].set_title('C. const + deviator')
axs[1].set_title('D. ada + deviator')
for a in axs:
    a.set_xlabel('iteration t')
    a.set_ylabel('strategy x_i(t)')
    a.set_ylim(-0.05, 1.05)
    a.grid(True, linestyle=':', alpha=0.5)
    a.legend(loc='best', fontsize=9)

# subplots_adjust instead of tight_layout (see note in previous cell).
fig.subplots_adjust(left=0.06, right=0.98, bottom=0.13, top=0.91, wspace=0.05)
plt.show()
```
